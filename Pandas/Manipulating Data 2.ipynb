# Manipulating Data 2

```python
# Column from series

seats_column = pd.Series([5, 5, 5, 5, 5])

# New Column called seats

car_sales["Seats"] = seats_column

car_sales
```

```python
car_sales["Seats"].fillna(5, inplace=True)
```

```python
car_sales
```

```python
# Column from Python List.

fuel_economy = [7.5, 9.2, 5.0, 9.6, 8.7]
# car_sales["Fuel per 100KM"] = fuel_economy

# # It Gave Error: ValueError: Length of values (5) does not match length of index (10)

# Fix: We need need data, so that no field is empty as it will throw error.
```

```python
fuel_economy = [7.5, 9.2, 5.0, 9.6, 8.7, 4.7, 7.6, 8.7, 3.0, 4.5]
car_sales["Fuel per 100KM"] = fuel_economy

car_sales
```

### Create a Column from Another Column

```python
car_sales["Total fuel used"] = car_sales["Odometer (KM)"] / 100 * car_sales["Fuel per 100KM"]
```

```python
car_sales
```

```python
# Renaming it but, it causes it to create a new Column.
car_sales["Total fuel used (L)"] = car_sales["Odometer (KM)"] / 100 * car_sales["Fuel per 100KM"]

car_sales
```

### Filtering the Data according to certain Conditions.

```python
# Boolean Indexing.

car_sales[car_sales["Make"] == "Toyota"]
```

```python
car_sales[car_sales["Odometer (KM)"] > 100000]
```

```python
car_sales
```

```python
pd.crosstab(car_sales["Make"], car_sales["Doors"])
```

```python
# Groupby
car_sales.groupby(["Make"]).mean(numeric_only=True)
```

```python
# Grouping by "White Toyotas" only.

car_sales.groupby((car_sales["Make"] == "Toyota") & (car_sales["Colour"] == "White")).groups
```

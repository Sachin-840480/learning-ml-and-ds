### .loc and .iloc

```python
# .loc

animals = pd.Series(['cat', 'dog', 'bird', 'panda', 'snake'])

animals
```

```python
animals = pd.Series(['cat', 'dog', 'bird', 'panda', 'snake'], index = [0, 3, 9, 8, 3])

animals
```

```python
# .loc refers to index.
animals.loc[3]
```

```python
animals.loc[9]
```

```python
car_sales
```

```python
car_sales.loc[3]
```

```python
# .iloc
```

```python
animals
```

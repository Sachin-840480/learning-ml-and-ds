# Dot Product vs Element Wise

```python
np.random.seed(0)
```

```python
mat1 = np.random.randint(10, size=(5,3))
mat2 = np.random.randint(10, size=(5,3))
```

```python
mat1.shape, mat2.shape
```

```python
mat1
```

```python
mat2
```

```python
# Element-wise multiplication (Hadamard product)

mat1 * mat2
```

```python
# Dot product 'np.dot' or '@'

# np.dot(mat1, mat2)
# Error: ValueError: shapes (5,3) and (5,3) not aligned: 3 (dim 1) != 5 (dim 0)

# mat1 @ mat2
# ValueError: matmul: Input operand 1 has a mismatch in its core dimension 0, with gufunc signature (n?,k),(k,m?)->(n?,m?) (size 5 is different from 3)
```

```python
# Transpose mat 1
mat1.T
```

```python
mat1.T.shape, mat2.shape
```

```python
mat1.shape, mat2.T.shape
```

```python
mat3 = np.dot(mat1, mat2.T)

mat3
```

# Dot product example (nut butter sales)

```python
import pandas as pd
```

```python
np.random.seed(0)

# Numbers of jars sold
sales_amount = np.random.randint(20, size=(5,3))
sales_amount
```

```python
# Create weekly_sales DataFrame

weekly_sales = pd.DataFrame(sales_amount, 
                            index=["Mon","Tues", "Wed", "Thurs", "Fri"],
                            columns=["Almond Butter", "Peanut Butter", "Cashew Butter"])

weekly_sales
```

```python
# Create prices array.

prices = np.array([10, 8, 12])

prices
```

```python
prices.shape
```

```python
# # Testing to see what it makes

# # Create butter_prices DataFrame
# butter_prices = pd.DataFrame(prices)

# butter_prices

# # So the thing is its creating a DF of "rows = 3 and col = 1", even though the value passed was "row = 1 and col = 3"
```

```python
# Create butter_prices DataFrame
butter_prices = pd.DataFrame(prices.reshape(1, 3),
                             index=["Price"],
                             columns=["Almond Butter", "Peanut Butter", "Cashew Butter"])                            

butter_prices
```

```python
prices.shape
```

```python
sales_amount.shape
```

```python
sales_amount.T.shape
```

```python
# The shapes arent aligned so it gives error.

# total_sales = prices.dot(sales_amount)

# Error: ValueError: shapes (3,) and (5,3) not aligned: 3 (dim 0) != 5 (dim 0)
```

```python
# Shapes aren't aligned, we need to transpose.

total_sales = prices.dot(sales_amount.T)

total_sales
```

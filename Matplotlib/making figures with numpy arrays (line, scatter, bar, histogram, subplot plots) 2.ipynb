# Making figures with NumPy arrays

We want:
* Line Plot.
* Scatter Plot.
* Bar Plot.
* Horizotal Bar Plot.
* Histogram.
* Subplot.

```python
import matplotlib.pyplot as plt
import numpy as np
```

### Line Plots

```python
# Create some data.

x = np.linspace(0, 10, 100)

x
```

```python
# Plot the data and create a line plot.

fig, ax = plt.subplots()
ax.plot(x, x**2);
```

### Scatter Plot

```python
# Use same data to make a scatter plot.

fig, ax = plt.subplots()
ax.scatter(x, np.exp(x))
plt.show()
```

```python
# Another scatter plot.

fig, ax = plt.subplots()
ax.scatter(x, np.sin(x))
plt.show()
```

### Bar Plots

```python
# Make a plot from dictionary.
nut_butter_prices = {"Almond Butter" : 10,
                     "Peanut Butter" : 8,
                     "Cashew Butter" : 12}

fig, ax = plt.subplots()
ax.bar(nut_butter_prices.keys(), nut_butter_prices.values())

# same thing but we can write it as 'height'backup.ipynb
# ax.bar(nut_butter_prices.keys(), height = nut_butter_prices.values())

ax.set(title="Satyam's Nut Butter Store",
       ylabel = "Price ($)")
plt.show()
```

```python
# Horizontal bar Pot

fig, ax = plt.subplots()
ax.barh(nut_butter_prices.keys(), nut_butter_prices.values())
```

### Histogram

```python
# Making Some Data for histogram and plot it.

x = np.random.randn(1000)

fig, ax = plt.subplots()
ax.hist(x)
```

### Subplots

#### Two options for subplots.

```python
# Subplot option 1

fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(nrows=2, 
                                             ncols=2, 
                                             figsize=(10, 5))

# Plot to each different axis.

ax1.plot(x, x/2)
ax2.scatter(np.random.random(10), np.random.random(10))
ax3.bar(nut_butter_prices.keys(), nut_butter_prices.values())
ax4.hist(np.random.randn(1000))
```

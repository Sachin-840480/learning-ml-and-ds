# Plotting from Pandas DataFrame 2

### Plotting on the Car_Sales

```python
car_sales
```

```python
# car_sales["Price"] = car_sales["Price"].str.replace('[\$\,]', '', regex=True)

# Or

car_sales["Price"] = car_sales["Price"].str.replace('\D', '', regex=True)

car_sales
```

```python
type(car_sales["Price"])
```

```python
# Remove last two zeros
car_sales["Price"] = car_sales["Price"].str.removesuffix("00")

# or # car_sales["Price"] = car_sales["Price"].str[:-2]

car_sales
```

```python
car_sales["Sales Date"] = pd.date_range("1/1/2020", periods=len(car_sales))

car_sales
```

```python
# Adding a Cummulative Sum Column.

car_sales["Total Sales"] = car_sales["Price"].cumsum()

car_sales
```

```python
print(car_sales["Price"].dtype)

print(type(car_sales["Price"][0]))
```

```python
# Fix: Change the datatype.

car_sales["Total Sales"] = car_sales["Price"].astype(int).cumsum()

car_sales
```

```python
# Let's plot the total sales.

car_sales.plot(x="Sales Date", y="Total Sales")
```

```python
# Reassigns price column to int.
# car_sales["Price"] = car_sales["Price"].astype(int)

# Plot scatter plot with price column as numeric.
car_sales.plot(x="Odometer (KM)", y="Price", kind = "scatter")
```

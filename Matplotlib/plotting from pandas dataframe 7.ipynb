```python
car_sales["Sales Date"] = pd.date_range("1/1/2020", periods=len(car_sales))

car_sales
```

```python
print(car_sales["Price"].dtype)

print(type(car_sales["Price"][0]))
```

```python
# Fix: Change the datatype.

car_sales["Total Sales"] = car_sales["Price"].astype(int).cumsum()

car_sales
```

# Plotting from Pandas Dataframe 5

### Which one should we use ? (pyplot vs matpoltlib OOP method ?)

* When plotting something quicky, it's okay to use the pyplot method.
* When plotting something more advanced, use the OO method.

```python
import matplotlib.pyplot as plt
```

```python
heart_disease.head()
```

```python
over_50 = heart_disease[heart_disease["age"] > 50]

over_50
```

```python
# Pyplot method.
over_50.plot(
    kind = "scatter",
    x = "age",
    y = "chol",
    c = "target",
    figsize=(10, 6)
)
```

```python
# OOP Method mixed with pyplot method. (Not Really True OOP)
fig, ax = plt.subplots(figsize=(10, 6))
over_50.plot(kind="scatter",
             x="age",
             y="chol",
             c="target",
             ax=ax)
```

```python
# OOP Method mixed with pyplot method, with adjusting the 'x-axis' limits. (Not Really True OOP)
fig, ax = plt.subplots(figsize=(10, 6))
over_50.plot(kind="scatter",
             x="age",
             y="chol",
             c="target",
             ax=ax)

ax.set_xlim([45, 100])
```

# Plotting from Pandas Dataframe 6

### True OOP Method.

```python
over_50.target.values
```

```python
# OOP Method from scratch.
fig, ax = plt.subplots(figsize=(10, 6))

# Plot the data.
scatter = ax.scatter(x = over_50["age"], 
                     y = over_50["chol"], 
                     c = over_50["target"])

# Customize the plot.
ax.set(title = "Heart Disease and Cholesterol Levels",
       xlabel = "Age",
       ylabel = "Cholesterol")

# Add a legend.
ax.legend(*scatter.legend_elements(), title="Target")

# Add a Horizontal Line.
ax.axhline(over_50["chol"].mean(),
           linestyle = "--")
```

# Plotting from Pandas Dataframe 7

```python
over_50
```

### Making a Subplot for:
* Heart Disease and Cholesterol Levels.
* Heart Disease and Max Heart Rate.

```python
# Subplot of chol, age, thalach.
fig, (ax1, ax2) = plt.subplots(nrows = 2, 
                               ncols = 1, 
                               figsize=(10, 10))

# Add data to ax1.
scatter = ax1.scatter(x = over_50["age"],
                      y = over_50["chol"],
                      c = over_50["target"])

# Customize ax1.
ax1.set(title = "Heart Disease and Cholesterol Levels",
        xlabel = "Age",
        ylabel = "Cholesterol")

# Add a legend to ax1.
ax1.legend(*scatter.legend_elements(), title = "Target")

# Add a mean-line.
ax1.axhline(y = over_50["chol"].mean(),
            linestyle = "--")

#---------------------------------------------------------------#

# Add data to ax2.
scatter = ax2.scatter(x = over_50["age"],
                      y = over_50["thalach"],
                      c = over_50["target"])

# Customize the ax2.
ax2.set(title = "Heart Disease and Max Heart Rate",
        xlabel = "Age",
        ylabel = "Max Heart Rate")

# Add a legend to ax2.
ax2.legend(*scatter.legend_elements(), title = "Target")

# Add a mean-line.
ax2.axhline(y = over_50["thalach"].mean(),
            linestyle = "--")
```

### Making these both subplots to share the same X-axis.

```python
# Using 'sharex'.

# Subplot of chol, age, thalach.
fig, (ax1, ax2) = plt.subplots(nrows = 2, 
                               ncols = 1, 
                               figsize=(10, 10),
                               sharex=True)

# Add data to ax1.
scatter = ax1.scatter(x = over_50["age"],
                      y = over_50["chol"],
                      c = over_50["target"])

# Customize ax1.
ax1.set(title = "Heart Disease and Cholesterol Levels",
        ylabel = "Cholesterol")

# Add a legend to ax1.
ax1.legend(*scatter.legend_elements(), title = "Target")

# Add a mean-line.
ax1.axhline(y = over_50["chol"].mean(),
            linestyle = "--")

#---------------------------------------------------------------#

# Add data to ax2.
scatter = ax2.scatter(x = over_50["age"],
                      y = over_50["thalach"],
                      c = over_50["target"])

# Customize the ax2.
ax2.set(title = "Heart Disease and Max Heart Rate",
        xlabel = "Age",
        ylabel = "Max Heart Rate")

# Add a legend to ax2.
ax2.legend(*scatter.legend_elements(), title = "Target")

# Add a mean-line.
ax2.axhline(y = over_50["thalach"].mean(),
            linestyle = "--")
```

### Adding a title to the Figure (both subplots.)

```python
# Subplot of chol, age, thalach.
fig, (ax1, ax2) = plt.subplots(nrows = 2, 
                               ncols = 1, 
                               figsize=(10, 10),
                               sharex=True)

# Add data to ax1.
scatter = ax1.scatter(x = over_50["age"],
                      y = over_50["chol"],
                      c = over_50["target"])

# Customize ax1.
ax1.set(title = "Heart Disease and Cholesterol Levels",
        ylabel = "Cholesterol")

# Add a legend to ax1.
ax1.legend(*scatter.legend_elements(), title = "Target")

# Add a mean-line.
ax1.axhline(y = over_50["chol"].mean(),
            linestyle = "--")

#---------------------------------------------------------------#

# Add data to ax2.
scatter = ax2.scatter(x = over_50["age"],
                      y = over_50["thalach"],
                      c = over_50["target"])

# Customize the ax2.
ax2.set(title = "Heart Disease and Max Heart Rate",
        xlabel = "Age",
        ylabel = "Max Heart Rate")

# Add a legend to ax2.
ax2.legend(*scatter.legend_elements(), title = "Target")

# Add a mean-line.
ax2.axhline(y = over_50["thalach"].mean(),
            linestyle = "--")

#---------------------------------------------------------------#

# Add a title to the figure.
fig.suptitle("Heart Disease Analysis", fontsize = 16, fontweight = 'bold')
```

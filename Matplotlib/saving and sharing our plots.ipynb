# Customizing Our Plots 2

```python
# Importing the Data.
heart_disease = pd.read_csv("./data/heart-disease.csv")

over_50 = heart_disease[heart_disease["age"] > 50]

over_50
```

### Trying various Colour Schemes.

```python
# Set the Style.
plt.style.use('seaborn-v0_8-whitegrid')

# OOP Method from scratch.
fig, ax = plt.subplots(figsize=(10, 6))

# Plot the data.
scatter = ax.scatter(x = over_50["age"], 
                     y = over_50["chol"], 
                     c = over_50["target"],
                     cmap="winter")       # This changes the colour scheme.

# Customize the plot.
ax.set(title = "Heart Disease and Cholesterol Levels",
       xlabel = "Age",
       ylabel = "Cholesterol")

# Add a legend.
ax.legend(*scatter.legend_elements(), title="Target")

# Add a Horizontal Line.
ax.axhline(over_50["chol"].mean(),
           linestyle = "--")
```

### Changing the "xlim" and "ylim" values.

```python
# Customizing the 'y' and 'x' axis limitations.

# Subplot of chol, age, thalach.
fig, (ax1, ax2) = plt.subplots(nrows = 2, 
                               ncols = 1, 
                               figsize=(10, 10),
                               sharex=True)

# # Change the 'ax1' "x-axis" & "y-axis" limits.
ax1.set_xlim([50, 80])
 
# Add data to ax1.
scatter = ax1.scatter(x = over_50["age"],
                      y = over_50["chol"],
                      c = over_50["target"],
                      cmap = "winter")

# Customize ax1.
ax1.set(title = "Heart Disease and Cholesterol Levels",
        ylabel = "Cholesterol")

# Add a legend to ax1.
ax1.legend(*scatter.legend_elements(), title = "Target")

# Add a mean-line.
ax1.axhline(y = over_50["chol"].mean(),
            linestyle = "--")

#---------------------------------------------------------------#

# Add data to ax2.
scatter = ax2.scatter(x = over_50["age"],
                      y = over_50["thalach"],
                      c = over_50["target"],
                      cmap = "winter")

# # Change the 'ax2' "x-axis" & "y-axis" limits.
ax2.set_xlim([50, 80])
ax2.set_ylim([60, 200])

# Customize the ax2.
ax2.set(title = "Heart Disease and Max Heart Rate",
        xlabel = "Age",
        ylabel = "Max Heart Rate")

# Add a legend to ax2.
ax2.legend(*scatter.legend_elements(), title = "Target")

# Add a mean-line.
ax2.axhline(y = over_50["thalach"].mean(),
            linestyle = "--")

#---------------------------------------------------------------#

# Add a title to the figure.
fig.suptitle("Heart Disease Analysis", fontsize = 16, fontweight = 'bold')
```

# Customizing Our Plots

### Customizing Matplotlib Plots and Getting Stylish

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
# See the different styles available.
plt.style.available

# print(len(plt.style.available))   # -> 28 Styles
```

```python
# Importing the datas.
car_sales = pd.read_csv("./data/car-sales.csv")

car_sales

# Converting the Number Column to Numeric
car_sales["Price"] = car_sales["Price"].str.replace('\D', '', regex=True)
car_sales["Price"] = car_sales["Price"].str.removesuffix("00").astype(int)
```

```python
car_sales.head()
```

```python
car_sales["Price"].plot()
```

```python
# Changing the style of the plots.

plt.style.use('seaborn-v0_8-whitegrid')
```

```python
# Creating some dummy data.
x = np.random.randn(10, 4)

x
```

```python
df = pd.DataFrame(x, columns=['a', 'b', 'c', 'd'])

df
```

```python
ax = df.plot(kind="bar")
type(ax)
```

```python
# Customize our plot with the set() method.

ax = df.plot(kind="bar")

# Add some labels and a title.
ax.set(title="Random Number Bar Graph from DataFrame",
       xlabel = "Row Number",
       ylabel = "Random Number")

# Make the legend visible.
ax.legend().set_visible(True)
```

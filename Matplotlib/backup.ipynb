## Matplotlib example workflow:-

```python
# 0. Import matplotlib and get it ready for plotting in Juypter.
import matplotlib.pyplot as plt

# 1. Prepare data.
x = [1, 2, 3, 4]
y = [11, 22, 33, 44]

# 2. Setup plot
fig, ax = plt.subplots(figsize=(10, 10))   # (width, height)

# 3. Plot the data.
ax.plot(x, y)

# 4. Customize plot.
ax.set(title="Simple Plot",
       xlabel="x-axis",
       ylabel="y-axis"
)

# 5. Save & show the figure.
fig.savefig("./images/sample-plot.png")
```

### Histogram

```python
# Making Some Data for histogram and plot it.

x = np.random.randn(1000)

fig, ax = plt.subplots()
ax.hist(x)
```

### Subplots

#### Two options for subplots.

```python
# Subplot option 1. (Recommended)

fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(nrows=2, 
                                             ncols=2, 
                                             figsize=(10, 5))

# Plot to each different axis.

ax1.plot(x, x/2)
ax2.scatter(np.random.random(10), np.random.random(10))
ax3.bar(nut_butter_prices.keys(), nut_butter_prices.values())
ax4.hist(np.random.randn(1000))
```

```python
# Subplots Option 2.
fig, ax = plt.subplots(nrows=2,
                       ncols=2,
                       figsize=(10, 5))

# Plot to each different axis (i.e index).
ax[0, 0].plot(x, x/2)
ax[0, 1].scatter(np.random.random(10), np.random.random(10))
ax[1, 0].bar(nut_butter_prices.keys(), nut_butter_prices.values())
ax[1, 1].hist(np.random.randn(1000))
```

```python
# Understanding the Axis of these Plots for both methods.

fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(nrows=2, ncols=2, figsize=(10, 5))

print(fig, (ax1, ax2), (ax3, ax4))

fig, ax = plt.subplots(nrows=2, ncols=2, figsize=(10, 5))

print(fig, ax)
```

# Plotting from Pandas DataFrame 5

### Which one should we use ? (pyplot vs matpoltlib OOP method ?)

* When plotting something quicky, it's okay to use the pyplot method.
* When plotting something more advanced, use the OO method.

```python
import matplotlib.pyplot as plt
```

```python
heart_disease.head()
```

```python
over_50 = heart_disease[heart_disease["age"] > 50]

over_50
```

```python
# Pyplot method.
over_50.plot(
    kind = "scatter",
    x = "age",
    y = "chol",
    c = "target",
    figsize=(10, 6)
)
```

```python
# OOP Method mixed with pyplot method. (Not Really True OOP)
fig, ax = plt.subplots(figsize=(10, 6))
over_50.plot(kind="scatter",
             x="age",
             y="chol",
             c="target",
             ax=ax)
```

```python
# OOP Method mixed with pyplot method, with adjusting the 'x-axis' limits. (Not Really True OOP)
fig, ax = plt.subplots(figsize=(10, 6))
over_50.plot(kind="scatter",
             x="age",
             y="chol",
             c="target",
             ax=ax)

ax.set_xlim([45, 100])
```

# Plotting from Pandas DataFrame 6

### True OOP Method.

```python
over_50.target.values
```

```python
# OOP Method from scratch.
fig, ax = plt.subplots(figsize=(10, 6))

# Plot the data.
scatter = ax.scatter(x = over_50["age"], 
                     y = over_50["chol"], 
                     c = over_50["target"])

# Customize the plot.
ax.set(title = "Heart Disease and Cholesterol Levels",
       xlabel = "Age",
       ylabel = "Cholesterol")

# Add a legend.
ax.legend(*scatter.legend_elements(), title="Target")

# Add a Horizontal Line.
ax.axhline(over_50["chol"].mean(),
           linestyle = "--")
```

### Making a Subplot for:
* Heart Disease and Cholesterol Levels.
* Heart Disease and Max Heart Rate.

```python
# Subplot of chol, age, thalach.
fig, (ax1, ax2) = plt.subplots(nrows = 2, 
                               ncols = 1, 
                               figsize=(10, 10))

# Add data to ax1.
scatter = ax1.scatter(x = over_50["age"],
                      y = over_50["chol"],
                      c = over_50["target"])

# Customize ax1.
ax1.set(title = "Heart Disease and Cholesterol Levels",
        xlabel = "Age",
        ylabel = "Cholesterol")

# Add a legend to ax1.
ax1.legend(*scatter.legend_elements(), title = "Target")

# Add a mean-line.
ax1.axhline(y = over_50["chol"].mean(),
            linestyle = "--")

#---------------------------------------------------------------#

# Add data to ax2.
scatter = ax2.scatter(x = over_50["age"],
                      y = over_50["thalach"],
                      c = over_50["target"])

# Customize the ax2.
ax2.set(title = "Heart Disease and Max Heart Rate",
        xlabel = "Age",
        ylabel = "Max Heart Rate")

# Add a legend to ax2.
ax2.legend(*scatter.legend_elements(), title = "Target")

# Add a mean-line.
ax2.axhline(y = over_50["thalach"].mean(),
            linestyle = "--")
```

### Making these both subplots to share the same X-axis.

```python
# Using 'sharex'.

# Subplot of chol, age, thalach.
fig, (ax1, ax2) = plt.subplots(nrows = 2, 
                               ncols = 1, 
                               figsize=(10, 10),
                               sharex=True)

# Add data to ax1.
scatter = ax1.scatter(x = over_50["age"],
                      y = over_50["chol"],
                      c = over_50["target"])

# Customize ax1.
ax1.set(title = "Heart Disease and Cholesterol Levels",
        ylabel = "Cholesterol")

# Add a legend to ax1.
ax1.legend(*scatter.legend_elements(), title = "Target")

# Add a mean-line.
ax1.axhline(y = over_50["chol"].mean(),
            linestyle = "--")

#---------------------------------------------------------------#

# Add data to ax2.
scatter = ax2.scatter(x = over_50["age"],
                      y = over_50["thalach"],
                      c = over_50["target"])

# Customize the ax2.
ax2.set(title = "Heart Disease and Max Heart Rate",
        xlabel = "Age",
        ylabel = "Max Heart Rate")

# Add a legend to ax2.
ax2.legend(*scatter.legend_elements(), title = "Target")

# Add a mean-line.
ax2.axhline(y = over_50["thalach"].mean(),
            linestyle = "--")
```

### Adding a title to the Figure (both subplots.)

```python
# Subplot of chol, age, thalach.
fig, (ax1, ax2) = plt.subplots(nrows = 2, 
                               ncols = 1, 
                               figsize=(10, 10),
                               sharex=True)

# Add data to ax1.
scatter = ax1.scatter(x = over_50["age"],
                      y = over_50["chol"],
                      c = over_50["target"])

# Customize ax1.
ax1.set(title = "Heart Disease and Cholesterol Levels",
        ylabel = "Cholesterol")

# Add a legend to ax1.
ax1.legend(*scatter.legend_elements(), title = "Target")

# Add a mean-line.
ax1.axhline(y = over_50["chol"].mean(),
            linestyle = "--")

#---------------------------------------------------------------#

# Add data to ax2.
scatter = ax2.scatter(x = over_50["age"],
                      y = over_50["thalach"],
                      c = over_50["target"])

# Customize the ax2.
ax2.set(title = "Heart Disease and Max Heart Rate",
        xlabel = "Age",
        ylabel = "Max Heart Rate")

# Add a legend to ax2.
ax2.legend(*scatter.legend_elements(), title = "Target")

# Add a mean-line.
ax2.axhline(y = over_50["thalach"].mean(),
            linestyle = "--")

#---------------------------------------------------------------#

# Add a title to the figure.
fig.suptitle("Heart Disease Analysis", fontsize = 16, fontweight = 'bold')
```

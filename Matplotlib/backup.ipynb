## Matplotlib example workflow:-

```python
# 0. Import matplotlib and get it ready for plotting in Juypter.
import matplotlib.pyplot as plt

# 1. Prepare data.
x = [1, 2, 3, 4]
y = [11, 22, 33, 44]

# 2. Setup plot
fig, ax = plt.subplots(figsize=(10, 10))   # (width, height)

# 3. Plot the data.
ax.plot(x, y)

# 4. Customize plot.
ax.set(title="Simple Plot",
       xlabel="x-axis",
       ylabel="y-axis"
)

# 5. Save & show the figure.
fig.savefig("./images/sample-plot.png")
```

### Line Plots

### Bar Plots

```python
# Make a plot from dictionary.
nut_butter_prices = {"Almond Butter" : 10,
                     "Peanut Butter" : 8,
                     "Cashew Butter" : 12}

fig, ax = plt.subplots()
ax.bar(nut_butter_prices.keys(), nut_butter_prices.values())

# same thing but we can write it as 'height'backup.ipynb
# ax.bar(nut_butter_prices.keys(), height = nut_butter_prices.values())

ax.set(title="Satyam's Nut Butter Store",
       ylabel = "Price ($)")
plt.show()
```

```python
# Horizontal bar Pot

fig, ax = plt.subplots()
ax.barh(nut_butter_prices.keys(), nut_butter_prices.values())
```

### Histogram

```python
# Making Some Data for histogram and plot it.

x = np.random.randn(1000)

fig, ax = plt.subplots()
ax.hist(x)
```

### Subplots

#### Two options for subplots.

```python
# Subplot option 1. (Recommended)

fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(nrows=2, 
                                             ncols=2, 
                                             figsize=(10, 5))

# Plot to each different axis.

ax1.plot(x, x/2)
ax2.scatter(np.random.random(10), np.random.random(10))
ax3.bar(nut_butter_prices.keys(), nut_butter_prices.values())
ax4.hist(np.random.randn(1000))
```

```python
# Subplots Option 2.
fig, ax = plt.subplots(nrows=2,
                       ncols=2,
                       figsize=(10, 5))

# Plot to each different axis (i.e index).
ax[0, 0].plot(x, x/2)
ax[0, 1].scatter(np.random.random(10), np.random.random(10))
ax[1, 0].bar(nut_butter_prices.keys(), nut_butter_prices.values())
ax[1, 1].hist(np.random.randn(1000))
```

```python
# Understanding the Axis of these Plots for both methods.

fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(nrows=2, ncols=2, figsize=(10, 5))

print(fig, (ax1, ax2), (ax3, ax4))

fig, ax = plt.subplots(nrows=2, ncols=2, figsize=(10, 5))

print(fig, ax)
```

# Plotting from Pandas Dataframe 5

### Which one should we use ? (pyplot vs matpotlib OOP method ?)

* When plotting something quicky, it's okay to use the pyplot method.
* When plotting something more advanced, use the OO method.

```python
import matplotlib.pyplot as plt
```

```python
heart_disease.head()
```

```python
over_50 = heart_disease[heart_disease["age"] > 50]

over_50
```

```python
# Pyplot method.
over_50.plot(
    kind = "scatter",
    x = "age",
    y = "chol",
    c = "target",
    figsize=(10, 6)
)
```

```python
# OOP Method. (Not Really True OOP)
fig, ax = plt.subplots(figsize=(10, 6))
over_50.plot(kind="scatter",
             x="age",
             y="chol",
             c="target",
             ax=ax)
```

```python
# OOP Method with adjusting the 'x-axis' limits. (Not Really True OOP)
fig, ax = plt.subplots(figsize=(10, 6))
over_50.plot(kind="scatter",
             x="age",
             y="chol",
             c="target",
             ax=ax)

ax.set_xlim([45, 100])
```

## Matplotlib example workflow:-

```python
# 0. Import matplotlib and get it ready for plotting in Juypter.
import matplotlib.pyplot as plt

# 1. Prepare data.
x = [1, 2, 3, 4]
y = [11, 22, 33, 44]

# 2. Setup plot
fig, ax = plt.subplots(figsize=(10, 10))   # (width, height)

# 3. Plot the data.
ax.plot(x, y)

# 4. Customize plot.
ax.set(title="Simple Plot",
       xlabel="x-axis",
       ylabel="y-axis"
)

# 5. Save & show the figure.
fig.savefig("./images/sample-plot.png")
```

# Making figures with NumPy arrays

We want:
* Line Plot.
* Scatter Plot.
* Bar Plot.
* Histogram.
* Subplot.

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Create some data.

x = np.linspace(0, 10, 100)

x
```

```python
# Plot the data and create a line plot.

fig, ax = plt.subplots()
ax.plot(x, x**2);
```

```python
# Use same data to make a scatter plot.

fig, ax = plt.subplots()
ax.scatter(x, np.exp(x))
plt.show()
```

```python
# Another scatter plot.

fig, ax = plt.subplots()
ax.scatter(x, np.sin(x))
plt.show()
```

```python
# Make a plot from dictionary.
nut_butter_prices = {"Almond Butter" : 10,
                     "Peanut Butter" : 8,
                     "Cashew Butter" : 12}

fig, ax = plt.subplots()
ax.bar(nut_butter_prices.keys(), nut_butter_prices.values())

# same thing but we can write it as 'height'backup.ipynb
# ax.bar(nut_butter_prices.keys(), height = nut_butter_prices.values())

ax.set(title="Satyam's Nut Butter Store",
       ylabel = "Price ($)")
plt.show()
```

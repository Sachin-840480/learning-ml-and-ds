# Making figures with NumPy arrays

We want:
* Line Plot.
* Scatter Plot.
* Bar Plot.
* Horizotal Bar Plot.
* Histogram.
* Subplot.

```python
import matplotlib.pyplot as plt
import numpy as np
```

### Line Plots

```python
# Create some data.

x = np.linspace(0, 10, 100)

x
```

```python
# Plot the data and create a line plot.

fig, ax = plt.subplots()
ax.plot(x, x**2);
```

### Scatter Plot

```python
# Use same data to make a scatter plot.

fig, ax = plt.subplots()
ax.scatter(x, np.exp(x))
plt.show()
```

```python
# Another scatter plot.

fig, ax = plt.subplots()
ax.scatter(x, np.sin(x))
plt.show()
```

### Bar Plots

```python
# Make a plot from dictionary.
nut_butter_prices = {"Almond Butter" : 10,
                     "Peanut Butter" : 8,
                     "Cashew Butter" : 12}

fig, ax = plt.subplots()
ax.bar(nut_butter_prices.keys(), nut_butter_prices.values())

# same thing but we can write it as 'height'backup.ipynb
# ax.bar(nut_butter_prices.keys(), height = nut_butter_prices.values())

ax.set(title="Satyam's Nut Butter Store",
       ylabel = "Price ($)")
plt.show()
```

```python
# Horizontal bar Pot

fig, ax = plt.subplots()
ax.barh(nut_butter_prices.keys(), nut_butter_prices.values())
```

### Histogram

```python
# Making Some Data for histogram and plot it.

x = np.random.randn(1000)

fig, ax = plt.subplots()
ax.hist(x)
```

### Subplots

#### Two options for subplots.

```python
# Subplot option 1. (Recommended)

fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(nrows=2, 
                                             ncols=2, 
                                             figsize=(10, 5))

# Plot to each different axis.

ax1.plot(x, x/2)
ax2.scatter(np.random.random(10), np.random.random(10))
ax3.bar(nut_butter_prices.keys(), nut_butter_prices.values())
ax4.hist(np.random.randn(1000))
```

```python
# Subplots Option 2.
fig, ax = plt.subplots(nrows=2,
                       ncols=2,
                       figsize=(10, 5))

# Plot to each different axis (i.e index).
ax[0, 0].plot(x, x/2)
ax[0, 1].scatter(np.random.random(10), np.random.random(10))
ax[1, 0].bar(nut_butter_prices.keys(), nut_butter_prices.values())
ax[1, 1].hist(np.random.randn(1000))
```

```python
# Understanding the Axis of these Plots for both methods.

fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(nrows=2, ncols=2, figsize=(10, 5))

print(fig, (ax1, ax2), (ax3, ax4))

fig, ax = plt.subplots(nrows=2, ncols=2, figsize=(10, 5))

print(fig, ax)
```

# Plotting from Pandas DataFrame 2.

### Plotting on the Car_Sales

```python
car_sales
```

```python
# car_sales["Price"] = car_sales["Price"].str.replace('[\$\,]', '', regex=True)

# Or

car_sales["Price"] = car_sales["Price"].str.replace('\D', '', regex=True)

car_sales
```

```python
type(car_sales["Price"])
```

```python
# Remove last two zeros
car_sales["Price"] = car_sales["Price"].str.removesuffix("00")

# or # car_sales["Price"] = car_sales["Price"].str[:-2]

car_sales
```

```python
car_sales["Sales Date"] = pd.date_range("1/1/2020", periods=len(car_sales))

car_sales
```

```python
# Adding a Cummulative Sum Column.

car_sales["Total Sales"] = car_sales["Price"].cumsum()

car_sales
```

```python
print(car_sales["Price"].dtype)

print(type(car_sales["Price"][0]))
```

```python
# Fix: Change the datatype.

car_sales["Total Sales"] = car_sales["Price"].astype(int).cumsum()

car_sales
```

```python
# Let's plot the total sales.

car_sales.plot(x="Sales Date", y="Total Sales")
```
